### import libraries and dataset

```python
import pandas as pd
import numpy as np

# load dataset

interactions = pd.read_csv('../data/raw/social_media_interactions_contaminated.csv')
```

### inspect

```python
# inspect dataset
print("First 20 rows: ")
display(interactions.head(20))

print("\nDataset Information: ")
display(interactions.info())

print("\nDataset Dimensions: ")
display(interactions.shape)

print("\nDataset Columns: ")
display(interactions.columns)

print("\nDataset Numerical Statistics: ")
display(interactions.describe())
```

### exact duplicates

```python
# inspect duplicates
duplicate_interactions = interactions.duplicated().sum()

print("Exact duplicate rows:", duplicate_interactions)

display(interactions[interactions.duplicated(keep=False)])
```

```python
# drop exact duplicate records
interactions_cleaned = interactions.drop_duplicates().copy()

# verify and check shape 
print("Duplicates remaining:", interactions_cleaned.duplicated().sum())
print("Dataset shape after removal of duplicates:", interactions_cleaned.shape)
```

removing exact duplicate records

the dataset initially contained 180 exact duplicate rows. these records contain identical values across all columns and therefore do not provide additional information for analysis so these exact duplicate rows were removed to prevent duplicated interactions from disproportionately influencing subsequent analysis and machine learning processes.

### missing, invalid, and inconsistent values

#### interaction id

```python
# check duplicate interaction ids
print("Duplicate InteractionIDs:", interactions_cleaned["InteractionID"].duplicated().sum())

# check number of records with unique interaction ids
print("\nNumber of Unique InteractionIDs:", interactions_cleaned["InteractionID"].nunique())
```

checking uniqueness of interaction ids

after removing exact duplicate rows, 20 `InteractionID` values were still duplicated. these records were inspected to determine whether they represented exact duplicates or potentially distinct interactions.

the duplicated records were found to contain differences in fields such as `InteractionDate` and `Sentiment`. since there was insufficient information to determine which record was correct, the records were retained rather than arbitrarily deleting one. these duplicated IDs are documented as data-quality anomalies.

```python
# verify
duplicate_interaction_ids = interactions_cleaned[interactions_cleaned.duplicated(subset="InteractionID", keep=False)].sort_values("InteractionID")

print("Records involved in duplicated InteractionIDs:", len(duplicate_interaction_ids))
display(duplicate_interaction_ids)
```

#### platform

```python
# since there were 311 missing values in the platform column that came out during initial inspection, inspect the partially cleaned dataset
print("Platform values:", interactions_cleaned["Platform"].value_counts(dropna=False))
print("\nUnique Platform values:", interactions_cleaned["Platform"].dropna().unique())
```

missing platform values

the platform olumn contains three consistently formatted categories: 'Twitter' 'Instagram' 'Facebook', no inconsistent platform labels were identified.

after removing exact duplicate records, 291 platform column values remained missing. since the correct platform cannot be reliably determined from the available data, the missing values were replaced with unknown rather than assigning an unsupported platform or removing the interaction records. this preserves the observations while explicitly identifying the missing information.

```python
# fill missing platform values with 'unknown'
interactions_cleaned["Platform"] = interactions_cleaned["Platform"].fillna("Unknown")

# verify
print("Missing Platform values:", interactions_cleaned["Platform"].isna().sum())
print("\nPlatform values:", interactions_cleaned["Platform"].value_counts())
```

#### sentiment

```python
# inspect column
print("Sentiment values:", interactions_cleaned["Sentiment"].value_counts(dropna=False))
print("\nUnique Sentiment values:", interactions_cleaned["Sentiment"].dropna().unique())
```

missing sentiment values
the sentiment column contains 5 consistently formatted categories: 'Neutral' 'Positive' 'Negative' 'Very Negative' 'Very Positive'. no inconsistent labels were identified among the existing values.

after removing exact duplicate records, 309 sentiment values remained missing. since sentiment cannot be reliably inferred from the available fields, the missing values were replaced with 'unknown' rather than assigning an unsupported sentiment category or removing the interaction records.

```python
# fill missing sentiment values with 'unknown'
interactions_cleaned["Sentiment"] = interactions_cleaned["Sentiment"].fillna('Unknown')

# verify
print("Missing Sentiment values:", interactions_cleaned["Sentiment"].isna().sum())
print("\nSentiment values:", interactions_cleaned["Sentiment"].value_counts())
```

#### interaction date

```python
# inspect interaction date formats
print("InteractionDate data type:", interactions_cleaned["InteractionDate"].dtype)

print("\nSample InteractionDate values:")
print(interactions_cleaned["InteractionDate"].head(20).to_string(index=False))

print("\nNumber of unique date values:", interactions_cleaned["InteractionDate"].nunique())

print("\nExamples containing '/':")
print(interactions_cleaned[interactions_cleaned["InteractionDate"].astype(str).str.contains("/", regex=False)]["InteractionDate"].head(20).to_string(index=False))
```

standardize interaction dates

the interaction date column was initially stored as a string and contained dates in multiple formats, including `yyyy-mm-dd` and `dd/mm/yyyy`.

the column was converted to a datetime data type using mixed-format parsing with `dayfirst=True`. invalid values were coerced to missing values so they could be identified during validation.

```python
# convert data type
interactions_cleaned["InteractionDate"] = pd.to_datetime(interactions_cleaned["InteractionDate"], format="mixed", dayfirst=True, errors="coerce")

# verify successful conversion
print("InteractionDate data type:", interactions_cleaned["InteractionDate"].dtype)

print("Missing InteractionDate values:", interactions_cleaned["InteractionDate"].isna().sum())

print("\nDate range:")
print("Earliest:", interactions_cleaned["InteractionDate"].min())
print("Latest:", interactions_cleaned["InteractionDate"].max())
```

the interaction date column was successfully converted from an object/string data type to datetime64[ns]. the mixed date formats were parsed correctly using day-first interpretation for slash-formatted dates.

validation confirmed that there are no missing or invalid interaction dates, and the resulting interaction dates range from July 1, 2023 to June 30, 2024.

### final validation

```python
print("FINAL SOCIAL MEDIA DATASET VALIDATION")
print("=" * 50)

print("\nDataset Shape:")
print(interactions_cleaned.shape)

print("\nData Types:")
print(interactions_cleaned.dtypes)

print("\nMissing Values:")
print(interactions_cleaned.isna().sum())

print("\nExact Duplicate Rows:")
print(interactions_cleaned.duplicated().sum())

print("\nDuplicate InteractionIDs:")
print(interactions_cleaned["InteractionID"].duplicated().sum())

print("\nPlatform Values:")
print(interactions_cleaned["Platform"].value_counts())

print("\nSentiment Values:")
print(interactions_cleaned["Sentiment"].value_counts())

print("\nInteractionDate Validation:")
print("Date data type:", interactions_cleaned["InteractionDate"].dtype)
print("Earliest date:", interactions_cleaned["InteractionDate"].min())
print("Latest date:", interactions_cleaned["InteractionDate"].max())
```

#### summary

after preprocessing, the social media interactions dataset contains 3,020 records and 6 columns. all missing values have been addressed, and no exact duplicate rows remain.

the platform and sentiment columns were checked for consistency. missing platform and sentiment values were labeled as 'unknown' because their correct values could not be reliably inferred from the available data.

The interaction date column was standardized to datetime64[ns], with no invalid or missing dates remaining.

the 20 duplicated interaction id values were investigated and retained because their associated records were not exact duplicates. these records are documented as data-quality anomalies rather than being arbitrarily removed.
