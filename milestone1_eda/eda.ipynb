## cross-dataset analysis

```python
# since the three datasets have different grains, we must not directly correlate individual transaction rows with individual customer demographic rows
## first, aggregate transactions and interactions to the customer level

transaction_customer_summary = (
    transactions
    .groupby("CustomerID")
    .agg(
        TotalSpend=("Amount", "sum"),
        TransactionCount=("TransactionID", "count"),
        AverageTransactionAmount=("Amount", "mean")
    )
    .reset_index()
)

interactions_customer_summary = (
    interactions
    .groupby("CustomerID")
    .agg(
        InteractionCount=("InteractionID", "count")
    )
    .reset_index()
)

customer_analysis = (
    demographics
    .merge(transaction_customer_summary, on="CustomerID", how="left")
    .merge(interactions_customer_summary, on="CustomerID", how="left")
)

customer_analysis[
    ["TotalSpend", "TransactionCount",
     "AverageTransactionAmount", "InteractionCount"]
] = customer_analysis[
    ["TotalSpend", "TransactionCount",
     "AverageTransactionAmount", "InteractionCount"]
].fillna(0)

display(customer_analysis.head())
```

### cross-dataset correlations

```python
customer_numeric = customer_analysis[
    [
        "Age",
        "TotalSpend",
        "TransactionCount",
        "AverageTransactionAmount",
        "InteractionCount"
    ]
]

correlation_matrix = customer_numeric.corr(method="pearson")

display(correlation_matrix)
```

The Age–TotalSpend relationship is essentially zero in this dataset.

The InteractionCount–TotalSpend relationship is also essentially zero.

The strong TransactionCount–TotalSpend relationship should not be presented as a discovery that transactions cause spending, because TotalSpend is mathematically constructed by summing transaction amounts and TransactionCount is derived from the same transaction records. That relationship is structurally expected.

### anova: product category vs amount

```python
from scipy.stats import f_oneway

product_groups = [
    group["Amount"].values
    for _, group in transactions.groupby("ProductCategory")
]

anova_result = f_oneway(*product_groups)

print("F-statistic:", anova_result.statistic)
print("p-value:", anova_result.pvalue)
```

At α = 0.05:

Since p = 0.2314 > 0.05, fail to reject H₀. The available data does not provide sufficient statistical evidence of a difference in mean transaction Amount across ProductCategory groups.
